```python
#variables and data types
Customer_Name = input("Enter the Customer name:")
Age = int(input("Enter the age of the customer:"))
Food_Items = input("Enter the food items that you required:")
Home_Delivery = bool(input("Need for home delivery:"))
Quantity = list(input("Enter the quantity of the food items:").split())
Price = list(input("Enter the price of each food item:").split())
#List
Menu = ["Biryani","Fried Rice","Soft Drink","Chicken 65","Muttoon Chukka","Ice cream"]
#Tuple
Delivery_Slots = ("1:00 PM - 3:00 PM", "7:00 PM - 9:00 PM", "10:00 PM - 11:00 PM")
#Set
Payment_Method = {"UPI", "COD", "NET BANKING", "CARD"}
#Dictionary
Order_Details = {"Customer name" : Customer_Name, "Food Items" : Food_Items, "Quantity" : Quantity, "Price" : Price}

#Customer Details
Customer_Name = Customer_Name.strip()
print("\n    Customer Details    ")
print("Customer Name: ",Customer_Name.upper())
print("Age: ",Age)
print("Home Delivery: ",Home_Delivery)

#Food Items Details
print("\n     Food Items     ")
print("Food Items: ",Food_Items.lower())
print("Length of the Food items: ",len(Food_Items))

#Menu Details
print("\n      Menu     ")
print("Available Menu: ",Menu)

#Delivery Slots
print("\n     Delivery Slots     ")
print("Delivery Slots: ", Delivery_Slots)

#Payment Method
print("\n     Payment Method     ")
print("Payment Method: ",Payment_Method)

#Order summary
print("\n     Order Summary of the Customer     ")
print("Order Summary: ",Order_Details)

```
